```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/data-export (1).csv")
```

```python
df.head()
```

```python
df.columns = df.iloc[0]
df= df.drop(index = 0). reset_index(drop =True)
df.columns = ["chanal group","DateHour", "Users","Sessions", "Engaged Sessions","Average engagement time per session", "Engaged sessions per user",	"Events per session","Engagement rate",	"Event count"]
```

```python
df.head()
```

```python
df.info()
```

#cleaning data and data validation for the above data set

```python
df["DateHour"] = pd.to_datetime(df["DateHour"], format="%Y%m%d%H", errors='coerce')
```

```python
df.info()
```

```python
df.head()
```

```python
numeric_cols = df.columns.drop(["chanal group","DateHour"])
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors = 'coerce')
df["Hour"] = df["DateHour"].dt.hour
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

#sessions and user over time

```python
sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(10,5))
df.groupby("DateHour")[["Sessions","Users"]].sum().plot(ax = plt.gca())
plt.title("Sessions and users over time")
plt.xlabel("DateHour")
plt.ylabel("count")
plt.show()
```

#total users by channel

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="chanal group", y="Users", estimator = np.sum,palette = "viridis")
plt.title("totle users by channel")
plt.xticks(rotation=45)
plt.show()
```

```python
df.head()
```

#Average engagment time by channel

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x= "chanal group",y ="Average engagement time per session", estimator =np.mean, palette = "magma")
plt.title("Avg engagement Time by Channel")
plt.xticks(rotation=45)
plt.show()
```

#Engagement Rate Distribution by channel

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="chanal group", y="Engagement rate", palette="coolwarm")
plt.title("Engagement Rate Distribution by channel")
plt.xticks(rotation=45)
plt.show()
```

#Engaged vs not engaged sessions

```python
df.head()
```

```python
session_df = df.groupby("chanal group")[["Sessions", "Engaged Sessions"]].sum().reset_index()
session_df["Non-Engaged"] = session_df["Sessions"] - session_df["Engaged Sessions"]
session_df_melted = session_df.melt(id_vars="chanal group", value_vars=["Engaged Sessions", "Non-Engaged"])

plt.figure(figsize=(8, 5))
sns.barplot(data=session_df_melted, x="chanal group", y="value", hue="variable")
plt.title("Engaged vs Non-Engaged Sessions")
plt.xticks(rotation=45)
plt.show()
```

#Traffic by hour and channel

```python
heatmap_data = df.groupby(["Hour","chanal group"])["Sessions"].sum().unstack().fillna(0)

plt.figure(figsize=(12,6))
sns.heatmap(heatmap_data,cmap="YlGnBu",linewidths=.5,annot=True, fmt=".0f")
plt.title("Traffic by hour and Channel")
plt.xlabel("Channel Hour")
plt.ylabel("Hour of Day")
plt.show()
```

#Engagement rate vs session over time

```python
df.head()
```

```python
df_plot = df.groupby("DateHour")[["Engagement rate","Sessions"]].mean().reset_index()

plt.figure(figsize=(10,5))
plt.plot(df_plot["DateHour"],df_plot["Engagement rate"], label = "Engagement rate", color ="green")
plt.plot(df_plot["DateHour"],df_plot["Sessions"], label="sessions",color="blue")
plt.title("engagement rate vs Sessions over time")
plt.xlabel("DateHour")
plt.grid(True)
plt.show()
```
